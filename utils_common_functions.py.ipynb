```python
from pyspark.sql.functions import current_timestamp

def add_ingestion_date(df):
    return df.withColumn("ingestion_date", current_timestamp())


def remove_duplicates(df, cols):
    return df.dropDuplicates(cols)


def filter_nulls(df, col_name):
    return df.filter(df[col_name].isNotNull())
```
